```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Reading dataset
df = pd.read_csv("./streaming_platform_eda_dataset.csv")

print(df.head())

sns.set_theme("notebook", style="whitegrid")



```

```python
# 1. What is the overall distribution of watch hours among users?
plt.figure(figsize=(12,6))

sns.histplot(data=df, x="Watch_Hours", bins=20, kde=True, color="steelblue")
 
plt.title("Distribution of Watch Hours", fontsize=18, fontweight="bold")

plt.xlabel("Watch Hours")
plt.ylabel("Number of Users")

plt.show()
```

```python
# 2. How are customer ratings distributed?
plt.figure(figsize=(12,6))

sns.histplot(data=df, x="Rating_Given", bins=20, kde=True, color="orange")
 
plt.title("Distribution of Rating", fontsize=18, fontweight="bold")

plt.xlabel("Rating Given")
plt.ylabel("Number of Users")
plt.show()
```

```python
# 3. Do Premium users watch significantly more content than Basic users?

plt.figure(figsize=(12,6))

sns.kdeplot(data=df, hue="Subscription_Type", x="Watch_Hours")
 
plt.title("Distribution of Rating", fontsize=18, fontweight="bold")

plt.xlabel("Number of Watch Hours")
plt.show()
```

```python
# 4. Does customer satisfaction differ between male and female users?

plt.figure(figsize=(12,6))

sns.kdeplot(data=df, hue="Gender", x="Rating_Given", fill=True)
 
plt.title("Distribution of Rating Between Male and Female", fontsize=18, fontweight="bold")

plt.xlabel("Ratings")
plt.show()
```

```python
# 5. Which genre attracts the largest audience?

plt.figure(figsize=(12,6))

sns.countplot(data=df, x="Favorite_Genre", order=df["Favorite_Genre"].value_counts().index, width=0.5, color="violet")
 
plt.title("Genre Comparison", fontsize=18, fontweight="bold")

plt.xlabel("Favorite Genre")
plt.ylabel("No of Audience")
plt.show()
```

```python
# 6. Which device dominates platform usage?

plt.figure(figsize=(8,6))

sns.countplot(data=df, x="Device", order=df["Device"].value_counts().index, width=0.4, color="cyan")
 
plt.title("Device Comparison", fontsize=14, fontweight="bold")

plt.xlabel("Device")
plt.ylabel("No of Audience")
plt.show()
```

```python
# 7. Which subscription plan is most popular?

plt.figure(figsize=(10,6))

sns.countplot(data=df, x="Subscription_Type", order=df["Subscription_Type"].value_counts().index, width=0.3, color="orange")
 
plt.title("Subscription Plan Comparison", fontsize=14, fontweight="bold")

plt.xlabel("Subscription Type")
plt.ylabel("No of Subscribers")
plt.show()
```

```python
# 8. Which subscription plan generates the highest average watch hours?

plt.figure(figsize=(8,6))

sns.barplot(data=df, x="Subscription_Type", y="Watch_Hours" , width=0.3)
 
plt.title("Subscription Plan Comparison to Watch Hours", fontsize=14, fontweight="bold")

plt.xlabel("Subscription Type")
plt.ylabel("Watch Hours")
plt.show()
```

```python
#  9. Which genre receives the highest average ratings?

plt.figure(figsize=(15,6))

sns.barplot(data=df, x="Favorite_Genre", y="Rating_Given", width=0.4)
 
plt.title("Ratings for Each Genre", fontsize=14, fontweight="bold")
plt.ylim([0,5])
plt.xlabel("Genre")
plt.ylabel("Ratings")
plt.show()
```

```python
# 10. Which city has the most engaged users?

plt.figure(figsize=(15,6))

sns.barplot(data=df, x="City", y="Watch_Hours", width=0.4)
 
plt.title("Watch Hours for Each City", fontsize=14, fontweight="bold")
plt.xlabel("City")
plt.ylabel("Watch_Hours")
plt.show()
```

```python
# 11. Do users who watch more content give higher ratings?

plt.figure(figsize=(15,6))

sns.scatterplot(data=df, x="Watch_Hours", y="Rating_Given")
 
plt.title("Correlation of Watch Hours and Rating Given", fontsize=14, fontweight="bold")
plt.xlabel("Watch Hours")
plt.ylabel("Rating")
plt.show()
```

```python
# 12. Does age influence viewing behavior?

plt.figure(figsize=(15,6))

sns.scatterplot(data=df, y="Watch_Hours", x="Age", color="green", marker="s")
 
plt.title("Correlation of Watch Hours and Age", fontsize=14, fontweight="bold")
plt.ylabel("Watch Hours")
plt.xlabel("Age")
plt.show()
```

```python
# 13. Do long-term members consume more content?

plt.figure(figsize=(15,6))

sns.scatterplot(data=df, y="Watch_Hours", x="Membership_Years", color="steelblue", marker="o")
 
plt.title("Correlation of Watch Hours and Membership Years", fontsize=14, fontweight="bold")
plt.ylabel("Watch Hours")
plt.xlabel("Membership Years")
plt.show()
```

```python
#  14. How does average watch time change as membership duration increases?

plt.figure(figsize=(15,6))

sns.lineplot(data=df, y="Watch_Hours", x="Membership_Years", color="steelblue", estimator="mean", errorbar=None)
 
plt.xlim([0,10])
plt.title(" Watch Hours Change Over Membership Years", fontsize=14, fontweight="bold")
plt.ylabel("Watch Hours")
plt.xlabel("Membership Years")
plt.show()
```

```python
#  15. How does average rating vary across different age groups?

plt.figure(figsize=(15,6))

sns.lineplot(data=df, x="Age", y="Rating_Given", color="green", marker="o")
 
plt.title("Average Rating Very Across Different Age Group", fontsize=14, fontweight="bold")
plt.ylabel("Ratings")
plt.xlabel("Age")
plt.show()
```

```python
#  16. Which subscription plan has the greatest variability in watch hours?

plt.figure(figsize=(15,7))

sns.boxplot(data=df, x="Subscription_Type", y="Watch_Hours", width=0.7)
 
plt.title("Variability of the Subscription Plan Over Watch Hours", fontsize=14, fontweight="bold")
plt.xlabel("Subscription Type")
plt.ylabel("Watch Hours")
plt.show()
```

```python
#  17. Which genre produces the widest range of ratings?
plt.figure(figsize=(15,7))

sns.boxplot(data=df, x="Favorite_Genre", y="Rating_Given", width=0.7, color="Orange")
 
plt.title("Range of Ratings Produced By Each Genre", fontsize=14, fontweight="bold")
plt.xlabel("Genre")
plt.ylabel("Ratings")
plt.show()
```

```python
#  18. Are there extreme binge-watchers on specific devices?
plt.figure(figsize=(15,6))

sns.boxplot(data=df, x="Device", y="Watch_Hours", width=0.7, color="green")
 
plt.title("Watcher On Each Device", fontsize=14, fontweight="bold")
plt.xlabel("Device")
plt.ylabel("Watchers")
plt.show()
```

```python
#  19. How is watch-hour distribution shaped within each subscription plan?

plt.figure(figsize=(15,7))

sns.violinplot(data=df, x="Subscription_Type", y="Watch_Hours", width=0.7)
 
plt.title("Watch Hour Distibution Over Subscription Plan", fontsize=14, fontweight="bold")
plt.xlabel("Subscription Type")
plt.ylabel("Watch Hours")
plt.show()
```

```python
#  20. Which genre shows the most concentrated ratings?

plt.figure(figsize=(15,7))

sns.violinplot(data=df, x="Favorite_Genre", y="Rating_Given", width=0.7)
 
plt.title("Ratings Distibution Over Each Genre", fontsize=14, fontweight="bold")
plt.xlabel("Genre")
plt.ylabel("Ratings")
plt.show()
```

```python
# 21. Which numeric features are most strongly related?
plt.figure(figsize=[15,7])
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".3f")

plt.title("Correlation Of The Numeric Values", fontsize=14, fontweight="bold")
plt.show()
```

```python
# 22. Which variables appear to influence ratings most?

filtered_corr = corr[['Rating_Given']].drop('Rating_Given')

plt.figure(figsize=(8, 6))
sns.heatmap(filtered_corr, annot=True, cmap='coolwarm', center=0, 
            fmt='.2f', linewidths=0.5, cbar_kws={'label': 'Correlation'})
plt.title('Filtered Correlation Heatmap\n(Only variables correlated with Rating Given)')
plt.ylabel('Variables')
plt.tight_layout()
plt.show()
```

```python
# 23. Are high-watch-hour users concentrated in Premium plans or spread across all plans?

plt.figure(figsize=(12,7))

sns.stripplot(data=df, x="Subscription_Type", y="Watch_Hours", jitter=True, alpha=0.5)
 
plt.title("Watch Hour Distibution Spread Over Subscription Type", fontsize=14, fontweight="bold")
plt.xlabel("Subscription Type")
plt.ylabel("Watch Hours")
plt.show()
```

```python
# 24. How are customer ratings distributed across genres at an individual level?

plt.figure(figsize=(15,7))

sns.stripplot(data=df, x="Favorite_Genre", y="Rating_Given", jitter=True, alpha=0.4)
 
plt.title("Rating Distibution Spread Over Individual Genre", fontsize=14, fontweight="bold")
plt.xlabel("Genre")
plt.ylabel("Ratings")
plt.show()
```

```python
# 25. Which device attracts the most consistent viewers?

plt.figure(figsize=(15,7))

sns.swarmplot(data=df, x="Device", y="Watch_Hours", color='green')
 
plt.title("Watch Hour Distribution Over Device", fontsize=14, fontweight="bold")
plt.xlabel("Device")
plt.ylabel("Watch Hours")
plt.show()
```

```python
# 26. Which subscription plan has the most tightly clustered ratings?

plt.figure(figsize=(15,8))

sns.swarmplot(data=df, x="Subscription_Type", y="Rating_Given", color='green')
 
plt.title("Rating Distribution For Each Subscription Type", fontsize=14, fontweight="bold")
plt.xlabel("Subscription Type")
plt.ylabel("Ratings")
plt.show()
```

```python
# 27. Perform a full relationship audit among all numerical variables.

numerical_vars = ['Age', 'Watch_Hours', 'Movies_Watched', 'Series_Watched', 
                  'Monthly_Fee', 'Membership_Years', 'Rating_Given']

sns.pairplot(df[numerical_vars])
plt.suptitle('Full Relationship Audit: All Numerical Variables', y=1.02)
plt.show()
```

```python
# 28. Investigate the relationship between watch hours and ratings while simultaneously examining their distributions.

sns.jointplot(data=df, x="Watch_Hours", y="Rating_Given", height=8, kind="scatter")
 
plt.title("Relationship Between Watch Hours And Ratings", fontsize=10, fontweight="bold")
plt.xlabel("Watch Hours")
plt.ylabel("Ratings")
plt.show()
```

```python
# 29. Can watch hours predict customer ratings?

plt.figure(figsize=(15,8))

sns.regplot(data=df, x="Watch_Hours", y="Rating_Given", ci=95, color="green")
 
plt.title("Regression Plot Between Watch Hours And Ratings", fontsize=14, fontweight="bold")
plt.xlabel("Watch Hours")
plt.ylabel("Ratings")
plt.show()
```

```python
#  30. Compare watch-hour distributions across subscription plans separately for each gender.

plt.figure(figsize=(15,8))

sns.catplot(data=df, y="Watch_Hours", x="Subscription_Type", col="Gender", kind="box", height=5, aspect=1)
 
plt.ylabel("Watch Hours")
plt.xlabel("Subscription type")
plt.show()
```
